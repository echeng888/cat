# ****This is a cat for Haaris. Hopefully he likes cats.****

This shows my proficiency in numpy and matplotlib. You can change the colour of the cat, by changing the colour hex code in the "head" and "ears" section.


```python
import numpy as np
import matplotlib.pyplot as plt

def circle(cx, cy, r, n=200):
    """Return x, y arrays for a circle centered at (cx, cy)."""
    t = np.linspace(0, 2 * np.pi, n)
    return cx + r * np.cos(t), cy + r * np.sin(t)

def ellipse(cx, cy, rx, ry, n=200, rot=0.0):
    """Return x, y arrays for a (optionally rotated) ellipse."""
    t = np.linspace(0, 2 * np.pi, n)
    x = rx * np.cos(t)
    y = ry * np.sin(t)
    # rotate
    xr = cx + x * np.cos(rot) - y * np.sin(rot)
    yr = cy + x * np.sin(rot) + y * np.cos(rot)
    return xr, yr

fig, ax = plt.subplots(figsize=(6, 6))

# --- Head ---
hx, hy = circle(0, 0, 1.0)
ax.fill(hx, hy, color="#9e9e9e", zorder=1) # change the colour code to change the colour of the cat

# --- Ears (triangles) ---
left_ear = np.array([[-0.85, 0.55], [-0.55, 1.25], [-0.25, 0.75]])
right_ear = np.array([[0.85, 0.55], [0.55, 1.25], [0.25, 0.75]])
ax.fill(left_ear[:, 0], left_ear[:, 1], color="#9e9e9e", zorder=1) # don't forget to change the ears too!
ax.fill(right_ear[:, 0], right_ear[:, 1], color="#9e9e9e", zorder=1) # don't forget to change the ears too!
# inner ears
ax.fill([-0.72, -0.55, -0.4], [0.62, 1.05, 0.72], color="#f7b7c8", zorder=2)
ax.fill([0.72, 0.55, 0.4], [0.62, 1.05, 0.72], color="#f7b7c8", zorder=2)

# --- Eyes ---
for ex in (-0.4, 0.4):
    x, y = ellipse(ex, 0.15, 0.16, 0.22)
    ax.fill(x, y, color="#2e7d32", zorder=2)          # green eye
    x, y = ellipse(ex, 0.15, 0.05, 0.20)
    ax.fill(x, y, color="black", zorder=3)            # slit pupil

# --- Nose ---
nose = np.array([[-0.12, -0.15], [0.12, -0.15], [0.0, -0.32]])
ax.fill(nose[:, 0], nose[:, 1], color="#f7b7c8", zorder=3)

# --- Mouth ---
mx = np.linspace(-0.25, 0.0, 50)
ax.plot(mx, -0.45 + 0.15 * np.cos(mx * 6), color="black", lw=1.5, zorder=3)
mx = np.linspace(0.0, 0.25, 50)
ax.plot(mx, -0.45 + 0.15 * np.cos(mx * 6), color="black", lw=1.5, zorder=3)

# --- Whiskers ---
for sign in (-1, 1):
    for dy in (0.0, 0.12, -0.12):
        ax.plot([sign * 0.2, sign * 0.95],
                [-0.2 + dy, -0.15 + dy * 1.5],
                color="black", lw=1, zorder=2)

ax.set_aspect("equal")
ax.axis("off")
ax.set_xlim(-1.6, 1.6)
ax.set_ylim(-1.6, 1.6)
plt.title("Cat for Haaris 🐱")
plt.show()
```
